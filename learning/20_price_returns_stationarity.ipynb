```python
import matplotlib.pyplot as plt
import yfinance as yf
import numpy as np
```

```python
data = yf.download("AAPL", start="2023-01-01", end="2025-01-01", auto_adjust=True)["Close"]
data.head()
```

```python
log_ret = np.log(data/data.shift(1))
log_ret.head()
```

```python
plt.figure(figsize=(15,8))
plt.subplot(121)
plt.plot(data[:300], "g")
plt.xlabel("Year")
plt.ylabel("Price")
plt.title("Non-stationarity  of prices")
plt.subplot(122)
plt.plot(log_ret[:300])
plt.title("Stationarity of log-returns")
plt.xlabel("Year")
plt.ylabel("Log returns")
plt.show()
```
