## **Modelling Simple Linear Regression**

The model $\hat{y} = \alpha + \beta x$, two unknowns $\alpha$ and slope $\beta $

Using **OLS** to minimize the total squared error between y and predicted $\hat{y}$.

RSS($\alpha , \beta$) = $\sum_{i=1}^{n}(y_i - \alpha - \beta x_i)^2$

Square the residuals so positive and negative errors don't cancel.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.DataFrame({
    "X":[1,2,3,4,5],
    "Y":[2,4,5,4,5]
})
data.head()
```

```python
x = data["X"]
y = data["Y"]
```

```python
x_mu = x.sum() / len(x)
y_mu = y.sum() / len(y)
```

Find $\beta$ 

$\hat{\beta} = \frac{\sum(x_1 - \bar{x})(y_i - \bar{y})}{\sum(x_i - \bar{x})^2} = \frac{Cov(x,y)}{Var(x)}$

```python
cov_x_y = 0
for i in range(len(data)):
    cov_x_y += (x[i] - x_mu)*(y[i] - y_mu)

var_x = 0
for i in range(len(x)):
    var_x += (x[i] - x_mu)**2
```

```python
min_beta = cov_x_y / var_x 
min_beta
```

Find alpha 

$\hat{\alpha} = \bar{y} - \hat{\beta}\bar{x}$
