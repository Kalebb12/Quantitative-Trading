## Two-dimensional dataset

```python
y = np.random.standard_normal((20, 2)).cumsum(axis=0)
```

```python
plt.figure(figsize=(10, 6))
plt.plot(y, 'ro')
plt.plot(y, lw=1.5)
plt.xlabel('index')
plt.ylabel('value')
plt.title('A simple plot')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.plot(y[:, 0], lw=1.5, label='1st')
plt.plot(y[:, 1], lw=1.5, label='2nd', color='lightgreen')
plt.plot(y , 'ro')
plt.xlabel('index')
plt.ylabel('value')
plt.title('A simple plot')
plt.legend(loc=0)
plt.show()
```

```python
y[:, 0] = y[:, 0] * 100 # Scaling 1st dataset
```

```python
plt.figure(figsize=(10, 6))
plt.plot(y[:, 0], lw=1.5, label='1st')
plt.plot(y[:, 1], lw=1.5, label='2nd')
plt.plot(y, 'ro')
plt.legend(loc=0)
plt.xlabel('index')
plt.ylabel('value')
plt.title('Simple plot')
plt.show()
```

```python
fig, axl = plt.subplots() # allows the creation of a second subplot that shares the same x-axis with the first subplot
plt.plot(y[:, 0], 'b', lw=1.5, label='1st')
plt.plot(y[:, 0], 'ro')
plt.legend(loc=8)
plt.xlabel('index')
plt.ylabel('value 1st')
plt.title('A simple plot')
ax2 = axl.twinx() # shares the x axis
plt.plot(y[:,1], 'g', lw=1.5, label='2nd')
plt.plot(y[:,1], 'ro')
plt.legend(loc=0)
plt.ylabel('value 2nd')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.subplot(211) # 2 rows 1 column and index position 1
plt.plot(y[:,0], lw=1.5, label='1st')
plt.plot(y[:,0], 'ro')
plt.legend(loc=0)
plt.ylabel('value')
plt.title('A simple plot')
plt.subplot(212) # 2 rows 1 column and index position 2
plt.plot(y[:,1], 'g', lw=1.5, label='2nd')
plt.plot(y[:,1], 'ro')
plt.legend(loc=0)
plt.ylabel('value')
plt.xlabel('index')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.subplot(121)
plt.plot(y[:, 0], lw=1.5, label='1st')
plt.plot(y[:, 0], 'ro')
plt.legend(loc=0)
plt.xlabel('index')
plt.ylabel('values')
plt.title('1st data set')
plt.subplot(122)
plt.bar(np.arange(len(y)), y[:, 1], width=0.5, color='g', label='2nd')
plt.legend(loc=0)
plt.xlabel('index')
plt.title('2nd data set')
plt.show()
```

```python
def func(x):
    return 0.5 * np.exp(x) + 1
a,b = 0.5, 1.5
x = np.linspace(0, 2)
y = func(x)
Ix = np.linspace(a, b)
Iy = func(Ix)
verts =[(a, 0)] + list(zip(Ix, Iy)) + [(b, 0)]
```

```python
from matplotlib.patches import Polygon
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
plt.plot(x, y, 'b', lw=2)
plt.ylim(ymin=0)
poly = Polygon(verts, facecolor='0.7', edgecolor='0.5')
ax.add_patch(poly)
plt.text(0.5 * (a + b), 1,r'$\int_a^b f(x)\mathrm{d}x$', horizontalalignment='center', fontsize=20)
plt.figtext(0.9, 0.075, '$x$')
plt.figtext(0.075, 0.9, '$f(x)$')
ax.set_xticks((a, b))
ax.set_xticklabels(('$a$', '$b$'))
plt.show()
```

## Static 3D plotting

```python
strike = np.linspace(50,150,24)
ttm = np.linspace(0.5, 2.5, 24)

strike, ttm = np.meshgrid(strike, ttm) # Two two-dimensional object created (grid)
```

```python
strike.shape
```

```python
strike[:2].round(1)
```

```python
iv = (strike - 100) ** 2 / (100 * strike) / ttm
iv[:5,:3]
```

```python
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(projection='3d')
surd = ax.plot_surface(strike, ttm, iv, rstride=2, cstride=2,
                       cmap=plt.cm.coolwarm, linewidth=0.5,
                       antialiased=True)
ax.set_xlabel('strike')
ax.set_ylabel('time-to-maturity')
ax.set_zlabel('implied volatility')
# fig.colorbar(surf, shrink=0.5, aspect=5)
plt.show()
```

```python
fig = plt.figure(figsize=(10, 6))
ax = fig.add_subplot(111, projection='3d')
ax.view_init(30, 60) # view angle
ax.scatter(strike,ttm,iv,zdir='z', s=25,c='b',marker='o')
ax.set_xlabel('strike')
ax.set_ylabel('time-to-maturity')
ax.set_zlabel('implied volatility')
plt.show()
```
