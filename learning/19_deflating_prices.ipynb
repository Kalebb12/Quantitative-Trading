```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../House Price and index.csv")
df.head()
```

2009 is the base year so CPI = 100 there. You're expressing every price as if it were in 2009 money.

```python
df["House Prices  (2009 Levels)"] = round((df["Nominal house prices"]/df["CPI (2009 levels)"] ) * 100)
df.head()
```

Same idea, but now 2018 is the reference year instead of 2009. You divide by the 2018 CPI value (123.6) instead of 100:

```python
cpi_2018 = 123.6
df["House Prices  (2018 Levels)"] = round((df["Nominal house prices"] / df["CPI (2009 levels)"]  ) * cpi_2018)
df
```
