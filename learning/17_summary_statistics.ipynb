## Calculaitng the summary statistics

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import skew, kurtosis
```

```python
data = pd.read_csv("../fund_manages.csv")
data
```

```python
plt.figure(figsize=(10, 6))
plt.plot(data["Year"], data["Risky Ricky"], label="Risky Ricky")
plt.plot(data["Year"], data["Safe Steve"], label="Safe Steve")
plt.title("Performance of the two managers by year")
plt.grid(alpha=0.4)
plt.legend()
plt.show()
```

```python
plt.title("Scatter plot on both managers")
plt.scatter(data["Risky Ricky"], data["Safe Steve"])
plt.show()
```

```python
mean = np.mean(data["Risky Ricky"])
std = np.std(data["Risky Ricky"])
# skew = skew(data["Risky Ricky"], bias=False)
kurt = kurtosis(data["Risky Ricky"], bias=False)

print(f"mean is {mean.round(2)}")
print(f"std is {std}")
# print(f"skew is {skew}")
print(f"kurt is {kurt.round(2)}")
```

```python
mean = np.mean(data["Safe Steve"])
std = np.std(data["Safe Steve"])
# skew = skew(data["Safe Steve"], bias=False)
kurt = kurtosis(data["Safe Steve"], bias=False)

print(f"mean is {mean.round(2)}")
print(f"std is {std}")
# print(f"skew is {skew}")
print(f"kurt is {kurt.round(2)}")
```
